#1. Write a Python program using the math module to find the factorial, square root, and
GCD of given numbers.

```python
import math
num=int(input("enter a number"))
num1=int(input("enter a number"))
squareroot=math.sqrt(num)
factor=math.factorial(num)
gratestcommondivisor=math.gcd(num,num1)
print("the square root is",squareroot)
print("the factorial is",factor)
print("the gratest common divisor is",gratestcommondivisor)
```

#3. Write a Python program using the datetime module to calculate a person's age from
their date of birth.

```python
import datetime
year=int(input("enter your year :"))
month=int(input("enter your month :"))
day = int(input("enter your day :"))
birth_date = datetime.date(year,month,day)
print("your date of birth is :",birth_date)
today_date=datetime.date.today()
print("today date is:",datetime.date.today().year)
age = today_date.year - birth_date.year
if (today_date.month,today_date.day) <(birth_date.month,birth_date.day):
  age -=1
print("your age is :",age)

```

#9. Create a user-defined module named calculator containing functions for addition,
subtraction, multiplication, and division. Import and use the module in another
program.
Hint: Create calculator.py with functions, then use import calculator in
main.py.

```python

%%writefile calculater.py

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Can't divide by zero"
    else:
        return a / b
```

```python

%%writefile main.py

import calculater

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition:", calculater.add(a, b))
print("Subtraction:", calculater.subtract(a, b))
print("Multiplication:", calculater.multiply(a, b))
print("Division:", calculater.divide(a, b))
```
