#1. Simple Shopping Cart

```python

list_products = [
    ("onion",50),
    ("milk",50),
    ("rice",80)
]
cart = []
print(f"the product we have is {list_products[0][0]},$ {list_products[0][1]} , {list_products[1][0]} ,$ {list_products[1][1]}, {list_products[2][0]}, $ {list_products[2][1]}")
items=input("enter what you want")
cart.append(items)
cart=items.split(",")
print("checking wherther the item are avilable or not :")
for i in list_products:
  if i in cart:
    print(f"the product {i} is available")
  else:
    print(f"the product {i} is not available")
product_remove = input("do you want to remove product(yes/no) ")
if product_remove == "yes":
  cart.remove(items)

print(f"your cart items are :{cart}")
print("the total bill is :")
total_bill=0
for i in cart:
  for j in list_products:
    if i in j:
      total_bill+=j[1]
print(f"the total is {total_bill}")


```

#7. Student Result System

```python
#creating a list
student_info = [
    ("Amit", 85, 78, 90),
    ("bibhu",99,99,99)
]

# Calculate total.
mark_info = []
for i in student_info:
  total_marks = 0
  for j in range(1,len(i)):
    total_marks = total_marks +i[j]
  # • Calculate percentage.
  persentage = total_marks/3

  ## • Assign grade.
  if persentage >= 80:
    grade = "A"
  elif persentage >=60:
    grade = "B"
  elif persentage  >=40:
    grade = "c"
  else:
    grade = "fail"
  # • Find pass/fail.
  if persentage >= 40:
    status = "pass"
  else:
    status = "fail"
  # • Display topper.

  print(f"the total mark of student :{i[0]} : {total_marks} : {persentage} : {grade} : {status}")
  mark_info.extend([(i[0],total_marks,persentage,grade,status)])
print(mark_info)
# • Display topper.

topper = mark_info[0]

for student in mark_info:
    if student[1] > topper[1]:
        topper = student

print("Topper:", topper[0])
print("Total marks:", topper[1])


```

#8. Movie Rating System
Concepts: List, tuple, loops
Store:
("Inception", 9)
("3 Idiots", 9.2)
("Dangal", 8.3)
Features:
• Add movie.
• Display movies.
• Find highest-rated movie.
• Search movie.


```python
movie_store = [("Inception", 9) ,
 ("3 Idiots", 9.2),
 ("Dangal", 8.3)
  ]
# • Add movie.
# • Display movies.
# • Find highest-rated movie.
# • Search movie.
print("""
1. Add movie press 1
2. Display movies press 2
3. Find highest-rated movie press 3
4. Search movie press 4
""")
choice = int(input("please give your choice "))
if choice == 1:
  # • Add movie.
  num_movie = int(input("please enter how many movie you want to enter :"))
  for i in range(num_movie):
    name = input("enter movie name :")
    rating = float(input("enter a your rating"))
    movie_store.extend([(name,rating)])
    print("your movie is inderted ")
  for i in movie_store:
    print(i)
elif choice ==2:
  # • Display movies.
  for i in movie_store:
    print(i)
elif choice ==3:
  # • Find highest-rated movie.
  ratings = int(input("please enter the reting you want to see :"))
  for i in movie_store:
    if i[1]>=ratings:
      print(i)
elif choice ==4:
  # • Search movie.
  movie_name = input("enter a movie name :")
  for i in movie_store:
    if i[0] == movie_name:
       print(i)

else:
  print("invalid choice")

# print(movie_store)
```

#10. Simple Billing System
Concepts: List of tuples, loops
Store:
("Pen", 10, 5)
("Notebook", 50, 2)
Where:
(Item, Price, Quantity)
Features:
• Calculate item-wise amount.
• Calculate total bill.
• Display final bill.

```python
store = [
    ("Pen", 10, 5),
    ("Notebook", 50, 2)
]
total_bill = 0
for i in store:
  price = i[1] * i[2]
  print(f"the price of {i[2]} {i[0]} is {price}")
  total_bill += price
print("the gross total is :",total_bill)
```
